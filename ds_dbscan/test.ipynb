```python
import numpy as np
from ds_dbscan import DSDBSCAN
from ds_dbscan.datasets import load_scans_csv, stack_polar

# load a recorded scan file (one of your Experiment 2 rooms)
path = r"C:\Users\herna\Documents\Hernan_Documentos\Thesis\OneDimensionKernelFunction\GitHub\OneDimensionKernelFunction\data_collection\Enviromentandexperiment\Experiment_2_real\2025.01.01_LargeRoom_BoxesInLine_03.csv"

scans = load_scans_csv(path)      # list of frames, each is (r, theta)
r, theta = scans[0]               # take the first frame

# now r and theta exist — cluster them
X = stack_polar(r, theta)         # builds the (n, 2) array of (range, angle)
labels = DSDBSCAN(eps=0.12, min_samples=3,
                  delta_theta=np.radians(1.0), s_obj=0.40).fit_predict(X)

print("points:", len(r))
print("clusters:", len(set(labels)) - (1 if -1 in labels else 0))
print("noise fraction:", np.mean(labels == -1))
```

```python
from sklearn.cluster import DBSCAN
from ds_dbscan.datasets import polar_to_xy

base = DBSCAN(eps=0.12, min_samples=3).fit_predict(polar_to_xy(r, theta))
print("DBSCAN     noise:", np.mean(base == -1), " clusters:", len(set(base)) - (1 if -1 in base else 0))
print("DS-DBSCAN  noise:", np.mean(labels == -1), " clusters:", len(set(labels)) - (1 if -1 in labels else 0))
```
